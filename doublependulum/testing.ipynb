```python
import numpy as np
```

```python
from doublependulum import DoublePendulum
```

```python
dp = DoublePendulum()
```

```python
from tqdm.notebook import tqdm
ts = [dp.t]
ys = [dp.transform_cartesian()]
xs = [dp.x]
F=0.0
for steps in tqdm(range(1000)):
    dp.step(F=F)
    ts.append(dp.t)
    ys.append(dp.transform_cartesian())
    xs.append(dp.x)
```

```python
from matplotlib import pyplot as plt
```

```python
plt.plot(ts, [y[0] for y in ys], label='Cart')
plt.plot(ts, [y[1][0] for y in ys], label='M1 (x)')
plt.plot(ts, [y[2][0] for y in ys], label='M2 (x)')

twx = plt.twinx()
twx.plot(ts, [y[1][1] for y in ys], c='C1', ls='--', label='M1 (y)')
twx.plot(ts, [y[2][1] for y in ys], c='C2', ls='--', label='M2 (y)')
```
